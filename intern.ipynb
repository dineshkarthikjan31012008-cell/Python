```python
# CSV Analysis

import pandas as pd

a = pd.read_csv("/content/sensor.csv")

print(a)

a.head()                       # Print first 5 rows

a.tail()                       # Print last 5 rows

a.head(50)                     # Print first 50 rows

a.tail(50)                     # Print last 50 rows

a.info()                       # Get CSV information
                                 # (columns, data types, non-null values, etc.)

a.describe()                   # Get statistical information
                                 # (count, mean, min, max, etc.)

a.shape                        # Get number of rows and columns

a.size                         # Get total number of data values

a.sample()                     # Get 1 random row

a.sample(5)                    # Get 5 random rows

a.sample(1, random_state=10)   # Get 1 random row consistently

a.columns                     # Show column names

a.index                        # Show row index

a.loc[25:51, ["Temperature", "Humidity"]]

a.iloc[25:50, [1, 2]]

a.at[4, "Pressure"]             # Get one exact value using row label and column name

a.iat[4, 3]                     # Get one exact value using row position and column position
```

```python
#boolean algebra!!!!!!!!!!!!!
import pandas as pd
a = pd.read_csv("/content/sensor.csv")
a.loc[a["Humidity"]<50]

a.loc[a["Temperature"]>25]

print(a.loc[(a["Temperature"]>25) & (a["Humidity"]<50)])

print(a.loc[(a["Temperature"]>25) | (a["Humidity"]<50)])

print(a.loc[~(a["Temperature"]>25) & (a["Humidity"]<50)])



```
